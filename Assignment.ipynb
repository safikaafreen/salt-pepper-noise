```python
import cv2
import numpy as np

img=cv2.imread("C:\\Users\\Administrator\\Pictures\\salt pepper noise image.jpeg")
med=cv2.medianBlur(img,3)
cv2.imshow("img",img)
cv2.waitKey(0)
cv2.imshow("BLUR-MED",med)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imwrite("clearimage.jpg",med)


MEDIAN=cv2.imread('clearimage.jpg')
bilat=cv2.bilateralFilter(MEDIAN,10,15,17)
cv2.imshow("img",img)
cv2.waitKey(0)
cv2.imshow("BLUR-BILAT",bilat)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
